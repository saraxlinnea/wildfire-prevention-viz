# Correlation Analysis (Exploratory)

Computes pairwise Pearson and Spearman correlations between chart variables
for overlapping year windows. **Not shown on the public visualization.**

See `../data/correlation-notes.md` for caveats (different geographies, fiscal vs
calendar year, short FS sample, correlation is not causation).

**Outputs:**
- `../data/correlation-by-window.csv`
- `../data/correlation-matrix.csv`

```python
import subprocess
from pathlib import Path

import pandas as pd

ROOT = Path("..")
subprocess.run(["python3", str(ROOT / "scripts/compute_correlations.py")], check=True)

pairs = pd.read_csv(ROOT / "data/correlation-by-window.csv")
matrix = pd.read_csv(ROOT / "data/correlation-matrix.csv", index_col=0)
pairs, matrix
```

```python
# Highlight strongest 2010-2025 climate+burn correlations (excluding prevention)
climate = pairs[pairs["window"] == "2010-2025 w/ VPD"].copy()
climate["abs_r"] = climate["pearson_r"].abs()
climate.sort_values("abs_r", ascending=False).head(10)
```
